```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
cnn = Sequential()
```

```python
cnn.add(Conv2D(32,(3,3), input_shape = (64,64,3), activation = 'relu'))
cnn.add(MaxPooling2D(pool_size = (2,2)))
cnn.add(Conv2D(32,(3,3), activation = 'relu'))
cnn.add(MaxPooling2D(pool_size = (2,2)))
cnn.add(Flatten())
```

```python
cnn.add(Dense(units = 64 ,activation = 'relu'))
cnn.add(Dense(units = 32 ,activation = 'relu'))
cnn.add(Dense(units = 16 ,activation = 'relu'))
cnn.add(Dense(units = 8 ,activation = 'relu'))
cnn.add(Dense(units = 4 ,activation = 'relu'))
cnn.add(Dense(units = 1 ,activation = 'sigmoid'))
cnn.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics=['accuracy'])
```

```python
train_datagen = ImageDataGenerator(
        rescale=1./255,
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True)

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
        r"C:\Users\Ajay\OneDrive\Desktop\CRAFTEDGE DATASET\CNN\Cat Dog dataset Test and Train\Training_set",
        target_size=(64,64),
        batch_size=32,
        class_mode='binary')

test_generator = test_datagen.flow_from_directory(
        r"C:\Users\Ajay\OneDrive\Desktop\CRAFTEDGE DATASET\CNN\Cat Dog dataset Test and Train\Test_set",
        target_size=(64,64),
        batch_size=32,
        class_mode='binary')

cnn.fit(
        train_generator,
        steps_per_epoch=251,
        epochs=20,
        validation_data=test_generator)
```
